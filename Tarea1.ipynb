# Método de bisección
En este código se encuentra la raíz del polinomio $f(x)=x^3+4x^2-10=0$ en el intervalo cerrado $[1,2]$, primero se comprueba que en efecto haya una raíz en el intervalo, es decir $f(1)*f(2)<0$, luego por el teorema del valor intermedio vamos aproximando la raíz hasta llegar al resultado con las cifras significativas que se soliciten

```python
import math as math
import numpy as np
import matplotlib.pyplot as plt
def funcion1(x):
  return x**3 + 4*pow(x,2) - 10
#Ingresa los datos de entrada y valida que sean el tipo de dato correcto
while True:
  try:
    a=float(input("Ingrese el límite inferior del intervalo cerrado [a,b]: "))
    break
  except:
    print("Por favor ingrese un número real")
while True:
  try:
    b=float(input("Ingrese el límite superior del intervalo cerrado [a,b]: "))
    break
  except:
    print("Por favor ingrese un número real")

#El sistema valida que en efecto haya una raiz en ese intervalo
if math.copysign(1,funcion1(a)) != math.copysign(1,funcion1(b)):
  xx=np.linspace(a-2,b+2,100)
  F1=funcion1(xx)
  plt.plot(xx,F1)
  plt.title("Función $f(x) = x^3 + 4x^2 - 10$")
  plt.grid(True)
  plt.axhline(0,color="black")
  plt.axvline(0,color="black")
  plt.show()
  print(f"Existe una raíz para la función en el intervalo [{a},{b}]")

  #guarda valores iniciales del error y del número de iteraciones
  while True:
    try:
      tol=float(input("Ingrese el número de tolerancia: "))
      break
    except:
      print("Ingrese un número real")
  while True:
    try:
      nmax=int(input("Ingresa el número máximo de iteraciones: "))
      break
    except:
      print("Ingrese un número entero")
  error=100
  niter=0

  #Método de Bisección

  #evaluar el primer valor medio
  m= a + (b - a)/2

  #Evaluación de la función en los puntos a, b y m
  fa=funcion1(a)
  fb=funcion1(b)
  fm=funcion1(m)
  print("")
  print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
  #guarda valores iniciales
  a0=a
  b0=b
  #ciclo iterativo
  while error > tol and niter < nmax:
    niter += 1
    m = a + (b - a)/2
    fm = funcion1(m)
    error = abs(b-a)
    print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter,a,fa,b,fb,m,fm,error))
    if np.sign(fa) == np.sign(fm):
      a=m
      fa=funcion1(a)
    else:
      b=m
      fb=funcion1(b)
  print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
else:
  xx=np.linspace(a-2,b+2,100)
  F1=funcion1(xx)
  plt.plot(xx,F1)
  plt.title("Función $f(x) = x^3 + 4x^2 - 10$")
  plt.grid(True)
  plt.axhline(0,color="black")
  plt.axvline(0,color="black")
  plt.show()
  print("No existe una raíz para la función en ese intervalo")
```
