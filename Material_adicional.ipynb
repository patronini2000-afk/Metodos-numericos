1) Demuestre que las siguientes ecuaciones tienen al menos una solución en los intervalos dados.

a) $\quad xcos(x) -2x^2 + 3x - 1, \qquad [0.2,0.3] \qquad y \qquad [1.2,1.3]$

SOLUCION: Consisdere la ecuación $\qquad f(x)=xcos(x) -2x^2 + 3x - 1$

Determinemos los valores para $f(0.2)$ y $f(0.3)$


```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return x*np.cos(x) - 2*(x**2) + 3*(x) - 1
print(f"f(0.2)={f(0.2)}")
print(f"f(0.3)={f(0.3)}")

xx=np.linspace(0,2,100)
F1=f(xx)
plt.plot(xx,F1)
plt.grid(True)
plt.axhline(0,color="black")
plt.axvline(0,color="black")
plt.show()
```

Puesto que $f \in C[0.2,0.3]$ y $f(0.2)< 0 < f(0.3)$, por el teroema del valor intermedio tenemos que existe $c \in \mathbb{R}$ tal que $f(c)=0$, es decir, la ecuación tiene al menos una solución.

Ahora determinemos los valores para $f(1.2)$ y $f(1.3)$

7. Demuestre que $f'(x)$ es $0$ al menos una vez en los intervalos dados

a) $f(x)=1-\mathrm{e}^x + (e-1)\sin (\frac{\pi}{2} x), \qquad [0,1]$

Note que $f(x)$ es continua pues la exponencial y la función seno son continuas y $f(x)$ es derivable.

Evaluemos la función en $0$ y $1$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return 1 - np.pow(np.exp(1),x) + (np.exp(1) - 1)*np.sin((np.pi*x)/2)
print(f"f(0)={f(0)}")
print(f"f(1)={f(1)}")

xx=np.linspace(0,1,100)
F1=f(xx)
plt.plot(xx,F1)
plt.grid(True)
plt.axhline(0,color="black")
plt.axvline(0,color="black")
plt.show()
```

$\therefore$ Por el teroema de Role $\exists c \in \mathbb{R} \mid f'(c)=0$

b) $f(x)= (x-1)\tan{x} + x \sin{(\pi x)}, \qquad [0,1]$

Note que al evualuar la función en ambos extremos se tiene que...

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return (x-1)*sp.tan(x) + x*sp.sin(sp.pi*x)
print(f"f(0)={f(0)}")
print(f"f(1)={f(1)}")

def s(x):
  return (x-1)*np.tan(x) + x*np.sin(np.pi*x)

xx=np.linspace(0,1,100)
F1=s(xx)
plt.plot(xx,F1)
plt.grid(True)
plt.axhline(0,color="black")
plt.axvline(0,color="black")
plt.show()
```

$\therefore$ Por el teroema de Role $\exists c \in \mathbb{R} \mid f'(c)=0$

c) $f(x)=x\sin{(\pi x)}-(x-2)\ln{(x)} \qquad [1,2]$

Como $x$ y $\sin{(\pi x)}$ son continuas en el intervalo descrito y $(x-2)$ y $\ln{(x)}$ también son continuas en dicho intervalo, se concluye que $f(x)$ es continua en $[1,2]$. Además $f(x)$ es diferenciable en $(1,2)$. Evaluemos los extremos de la función:




```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return x*sp.sin(sp.pi*x)-(x-2)*sp.log(x)
print(f"f(1)={f(1)}")
print(f"f(2)={f(2)}")

def s(x):
  return x*np.sin(np.pi*x)-(x-2)*np.log(x)

xx=np.linspace(0,2,100)
F1=s(xx)
plt.plot(xx,F1)
plt.grid(True)
plt.axhline(0,color="black")
plt.axvline(0,color="black")
plt.show()
```

Como $f(1)=f(2)$ por el Teoremad e Role $\exists c \in \mathbb{R} \mid f'(c)=0$

d) $f(x)=(x-2)\sin{(x)} \ln{(x+2)}, \qquad [-1,3]$

Como la función es contunua en el intervalo dado y $f(x)$ es diferenciable en $(-1,3)$; y al evaluar en los extremos tenemos que...

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return (x-2)*sp.sin(x)*sp.log(x+2)
print(f"f(-1)={f(-1)}")
print(f"f(3)={f(3)}")

def s(x):
  return (x-2)*np.sin(x)*np.log(x+2)

xx=np.linspace(-1,3,100)
F1=s(xx)
plt.plot(xx,F1)
plt.grid(True)
plt.axhline(0,color="black")
plt.axvline(0,color="black")
plt.show()
```

$f(-1)=0$, pero $f(3)\neq 0$, no obsatnte $\sin{(0)}=0$ y toda la función se hace $0$, i.e. para el intervalo $[-1,0]$ se sumple el teorema de Role

$\therefore \exists c \in \mathbb{R} \mid f'(c)=0$ y se se cumple para ese intervalo, también se cumple que existe al  menos un $f'(c)=0$ en el intervalo $[-1,3]$
