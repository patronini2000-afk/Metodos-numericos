# Método de Neville

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from tabulate import tabulate
```

## Damos un espacio para definir la función en caso de que exista una

```python
def f(t):
    return pow(3,t)
```

## Le pedimos al usuario que ingrese el valor para x y el número de nodos a utilizar

```python
while True:
    try:
        x=float(input("Ingrese el valor para x: "))
        break
    except:
        print("Ingrese un número real")

while True:
    try:
        nodos=int(input("Ingrese el número de nodos a utilizar: "))
        break
    except:
        print("Ingrese un número entero")
```

## Pedimos al usuario que ingrese los nodos y los guardamos en una lista para mandarlos a llamar más adelante

```python
vector_nodos=[]
for i in range(nodos):
    while True:
        try:
            vector_nodos.append(float(input(f"Ingrese el valor del nodo x{i}: ")))
            break
        except:
            print("Ingrese un número real")
```

## Generamos la matriz para guardar los $Q_{i,j}$; además le damos la opción al usuario de ingresar manualmente los f(x), es decir los $Q_{i,0}$ para i=1,2,...,n; o bien usar la función que se definió anteriormente. Note que el método está condensado en la útima linea de código

```python
matriz=[]
while True:
    try:
        op=int(input(f"Escriba 1 si desea ingresar los Q_i,0 de forma manual \\ escriba 2 si ya introdujo la función"))
        if op==1 or op==2:
            break 
        else:
            print("Seleccione una de las opciones")
    except:
        print("Ingrese un número entero")
for i in range(nodos):
    matriz.append([])
for i in range(nodos):
    for j in range(nodos):
        if j==0:
            if op==1:
                while True:
                    try:
                        matriz[i].append(float(input(f"Ingrese Q_({i},0)")))
                        break
                    except:
                        print("Ingrese un número real")
            elif op==2:
                matriz[i].append(float(f(vector_nodos[i])))
        elif j>0:
            if i<j:
                matriz[i].append(0)
            else:
                matriz[i].append(((x- vector_nodos[i-j])*matriz[i][j-1]-(x- vector_nodos[i])*matriz[i-1][j-1])/(vector_nodos[i]-vector_nodos[i-j]))
```

## Codensamos la tabala de los $Q_{i,j}$ y de la lista de los nodos en una sola tabla

```python
matrizfinal=[]
for i in range(nodos):
    matrizfinal.append([])
for i in range(nodos):
    for j in range(nodos+1):
        if j==0:
            matrizfinal[i].append(vector_nodos[i])
        else:
            matrizfinal[i].append(matriz[i][j-1])
```

## Agregamos una cabezera y generamos la tabla

```python
cabecera=[]
for i in range(nodos+1):
    if i==0:
        cabecera.append("xi")
    else:
        cabecera.append(f"y{i-1}")
df = pd.DataFrame(matrizfinal)
print(tabulate(df, headers=cabecera, tablefmt='fancy_grid', showindex=False))
```
