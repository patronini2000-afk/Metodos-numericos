```python
import numpy as np
import matplotlib.pyplot as plt
```

Vamos a grafiar la finción $f(x)=e^{x^2}$

```python
def f(x):
    return np.exp(x**2)
print(f"f(1)={f(1)}")

xx=np.linspace(0,1,100)
F1=f(xx)
plt.plot(xx,F1)
plt.title("Función $f(x) = e^{x^2}$")
plt.grid(True)
plt.axhline(0,color="black")
plt.axvline(0,color="black")
plt.show()
```
